## Part 1 — Working with Python Lists

### Step 1: Create a mixed-data list
Create a list with integers, strings, and floating-point numbers.

```python
my_list = [25, "Hello", 3.14, "Python", 100, 7.89, "AI", 42, "Data", 2.718]
print(my_list)
```

### Step 2: Positive indexing
Select list elements by their positions, starting at zero.

```python
positive_indexing = [my_list[0], my_list[3], my_list[6]]
print(positive_indexing)
```

### Step 3: Negative indexing
Use negative positions to count backward from the end.

```python
negative_indexing = [my_list[-1], my_list[-4], my_list[-7]]
print(negative_indexing)
```

### Step 4: Slicing
Select indices 2 through 6; the ending index 7 is excluded.

```python
sublist = my_list[2:7]
print(sublist)
```

### Step 5: Append and remove
Add a new item, then remove an existing item.

```python
my_list.append("New Element")
print(my_list)

my_list.remove("Hello")
print(my_list)
```

### Step 6: Copy items using a for loop
Append each element from the original list into a second list.

```python
new_list = []
for n in my_list:
    new_list.append(n)
print(new_list)
```

### Step 7: Repeated list elements
Reset the list to demonstrate repeated values.

```python
my_list = [25, 25, 35, 35, "Hello", "hello", 3.14, "Python", 100, 7.89, "AI", 42, "Data", 2.718]
print(my_list)
```

### Step 8: Use a set to remove duplicates
A set keeps unique elements. Capitalization matters, and sets do not guarantee display order.

```python
my_set = set()
for n in my_list:
    my_set.add(n)
print(my_set)
```

### Step 9: Negative indexing after resetting the list
The positions have changed because `my_list` was reassigned in Step 7.

```python
negative_indexing = [my_list[-1], my_list[-4], my_list[-7]]
print(negative_indexing)
```

## Part 2 — Feature Requests and Loops

### Step 10: Create a feature-request list
Use a list to store proposed computer features.

```python
possibleFeatures = ["storage", "space", "memory", "cpu"]
featureRequests = [
    "i wish i had more storage",
    "it would be good to have more cpu",
    "i wish i had more memory",
    "memory and cpu should be more",
    "i need more storage",
    "i need more cpu",
    "i wish i had more memory",
    "i wish i had more space"
]
print(possibleFeatures)
print(featureRequests)
```

### Step 11: Search feature requests with a loop
For each request, print which feature words appear in the text.

```python
for everyrequest in featureRequests:
    for feature in possibleFeatures:
        if feature in everyrequest:
            print(feature, ":", everyrequest)
```

```python
with open("Sample_dataO.csv", "r", newline="") as csvFile:
    csvReader = csv.reader(csvFile)
    print(type(csvReader))
    for line in csvReader:
        print(line)
```

### Step 14: Distinguish header and rows
The first row is treated as the column header in this sample file.

```python
with open("Sample_dataO.csv", "r", newline="") as csvFile:
    csvReader = csv.reader(csvFile)
    for count, line in enumerate(csvReader):
        if count == 0:
            print("Header:", line)
        else:
            print("Row:", line)
```

### Step 19: Create, append to, and read a text file
Use `w` to create, `a` to append, and `r` to read.

```python
with open("pytestcr.txt", "w") as a:
    for i in range(10):
        a.write(f"This is line {i+1}\n")

with open("pytestcr.txt", "a") as a:
    for i in range(5):
        a.write(f"This is appended line {i+1}\n")

with open("pytestcr.txt", "r") as a:
    print(a.read())
```

### Programmatic Representation: Nested For Loops Feature Matching
Below is the Python code using `networkx` and `matplotlib` to dynamically draw the matching relationship diagram.

```python
import matplotlib.pyplot as plt
import networkx as nx

# Define our categories and items
possible_features = ["storage", "space", "memory", "cpu"]
requests = [
    "i wish i had more storage",
    "it would be good to have more cpu",
    "i wish i had more memory",
    "memory and cpu should be more",
    "i need more storage",
    "i need more cpu",
    "i wish i had more memory",
    "i wish i had more space"
]

# Create a directed graph to show relations
G = nx.DiGraph()

# Add nodes
for f in possible_features:
    G.add_node(f, bipartite=0)
for r in requests:
    G.add_node(r, bipartite=1)

# Create edges based on nested loop logic
for r in requests:
    for f in possible_features:
        if f in r:
            G.add_edge(r, f)

# Setup positions for a bipartite-like layout structure
pos = {}
# Position requests on the left
for i, r in enumerate(requests):
    pos[r] = (1, len(requests) - i)
# Position possible features on the right
for i, f in enumerate(possible_features):
    pos[f] = (2, (len(requests) / len(possible_features)) * (len(possible_features) - i) + 1.5)

# Draw the diagram
plt.figure(figsize=(12, 8))
plt.title("Nested For Loop matching: Requests -> Target Features", fontsize=14, fontweight='bold', pad=20)

nx.draw_networkx_nodes(G, pos, nodelist=requests, node_color="lightblue", node_size=3000, node_shape="s")
x_features = nx.draw_networkx_nodes(G, pos, nodelist=possible_features, node_color="orange", node_size=3000, node_shape="o")

nx.draw_networkx_edges(G, pos, edgelist=G.edges(), edge_color="gray", width=2, arrowstyle="->", arrowsize=15)

# Shift label coordinates slightly for clarity
labels_requests = {r: r for r in requests}
labels_features = {f: f.upper() for f in possible_features}

nx.draw_networkx_labels(G, pos, labels=labels_requests, font_size=8, font_weight="bold")
nx.draw_networkx_labels(G, pos, labels=labels_features, font_size=10, font_weight="bold")

plt.axis("off")
plt.tight_layout()
plt.show()
```
