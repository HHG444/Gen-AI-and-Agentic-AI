```python
"""
Lesson 04 - Guided Practice
Practical Application of AI-Powered Code Generation Tools

Number Guessing Game

This program creates a simple number-guessing game.
The computer generates a random number between 1 and 100,
and the player has 5 attempts to guess the correct number.

The comments and docstrings demonstrate how descriptive prompts
can guide GitHub Copilot when generating Python code.
"""

# ---------------------------------------------------------
# STEP 1: SET UP THE PROJECT
# ---------------------------------------------------------

# This script implements a number-guessing game using Python.


# ---------------------------------------------------------
# STEP 2: GENERATE A RANDOM NUMBER
# ---------------------------------------------------------

# Importing random module to generate a random number
import random


def generate_random_number():
    """
    Generates a random number between 1 and 100.

    Returns:
        int: A randomly selected number.
    """

    # randint includes both 1 and 100
    random_number = random.randint(1, 100)

    return random_number


# ---------------------------------------------------------
# STEP 3: IMPLEMENT USER INPUT HANDLING
# ---------------------------------------------------------

def get_user_guess():
    """
    Prompts the user to enter a guess and validates the input.

    Returns:
        int: The user's valid guess.
    """

    while True:

        try:
            # Ask the player to enter a number
            guess = int(input("Enter your guess (1-100): "))

            # Make sure the guess is inside the valid range
            if 1 <= guess <= 100:
                return guess

            # The value is a number, but it is outside the valid range
            print("Please enter a number between 1 and 100.")

        except ValueError:
            # This happens when the player enters something
            # that cannot be converted to an integer
            print("Invalid input. Please enter a whole number.")


# ---------------------------------------------------------
# STEP 4: WRITE THE GAME LOGIC
# ---------------------------------------------------------

def play_game():
    """
    Runs the number guessing game.

    The user attempts to guess the randomly generated
    number within 5 tries.
    """

    # Generate the secret number
    secret_number = generate_random_number()

    # Set the maximum number of attempts
    max_attempts = 5

    # Welcome message
    print("=" * 50)
    print("          NUMBER GUESSING GAME")
    print("=" * 50)

    print("\nI'm thinking of a number between 1 and 100.")
    print(f"You have {max_attempts} attempts to guess it.")

    # Loop through the five available attempts
    for attempt in range(1, max_attempts + 1):

        print()
        print(f"Attempt {attempt} of {max_attempts}")

        # Get a valid guess from the player
        guess = get_user_guess()

        # Check whether the guess is correct
        if guess == secret_number:

            print("\nCorrect!")
            print(f"You guessed the number in {attempt} attempt(s).")

            # Return ends the game immediately
            return

        # Tell the player when the guess is too low
        elif guess < secret_number:

            print("Too low. Try a higher number.")

        # Tell the player when the guess is too high
        else:

            print("Too high. Try a lower number.")

        # Calculate how many attempts remain
        attempts_left = max_attempts - attempt

        if attempts_left > 0:
            print(f"You have {attempts_left} attempt(s) left.")

    # This code runs only if the player does not guess
    # the correct number within five attempts
    print("\nGame over!")
    print(f"The correct number was {secret_number}.")


# ---------------------------------------------------------
# STEP 5: TEST / RUN THE GAME
# ---------------------------------------------------------

# Run play_game() only when this Python file is executed
# directly. This is a standard Python best practice.
if __name__ == "__main__":
    play_game()
```
