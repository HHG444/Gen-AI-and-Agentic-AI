```python
print('--- Checking References ---')
x_ref = 34
y_ref = x_ref
print('x = ', x_ref, ';  id of x = ', id(x_ref))
print('y = ', y_ref, ';  id of y = ', id(y_ref))

x_ref = 65
print('x = ', x_ref, ';  id of x = ', id(x_ref))
print('y = ', y_ref, ';  id of y = ', id(y_ref))

y_ref = 78
print('x = ', x_ref, ';  id of x = ', id(x_ref))
print('y = ', y_ref, ';  id of y = ', id(y_ref))

print('\n--- Arithmetic Operators ---')
x_calc = 50
y_calc = 4
print('*'*10)
print('Sum of x & y                   : ', x_calc + y_calc)
print('Difference between x & y       : ', x_calc - y_calc)
print('Product of x & y               : ', x_calc * y_calc)
print('Division of x by y             : ', x_calc / y_calc)
print('*'*10)
print('Remainder of x divided by y    : ', x_calc % y_calc)
print('Floor Division x divided by y  : ', x_calc // y_calc)
print('Exponential value for x raised to power y : ', x_calc ** y_calc)

print('\n--- Arithmetic Assignment Operators ---')
a_val = 10
print('Present value of a = ', a_val)
a_val = a_val + 1
print('Value after a = a+1:', a_val)
a_val += 5
print('Updated value of a += 5 = ', a_val)
a_val += 5
print('Updated value of a += 5 again = ', a_val)

print('\n--- Comparison & Logical Operators ---')
a_comp, b_comp = 20, 20
print('a == b  : ', a_comp == b_comp)
a_comp, b_comp = 20, 30
print('a != b  : ', a_comp != b_comp)
print('a is b  : ', a_comp is b_comp)
```
