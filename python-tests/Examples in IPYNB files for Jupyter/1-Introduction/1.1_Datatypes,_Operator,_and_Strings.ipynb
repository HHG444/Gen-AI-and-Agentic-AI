```python
permitted_cities = ['New York', 'Los Angeles', 'Chicago', 'Zurich']
name = input('a name: ')
age = int(input('age: '))
city = input('city: ')

for everyelement in permitted_cities:
    if city == everyelement.lower() or city == everyelement.upper():
        if age > 30:
            print(f'{name}, age {age} from {city} is allowed to attend the conference.')
else:
    print('enter your city name in all lower or upper case')
```
